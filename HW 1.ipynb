```python
# Question 1

import numpy as np

a = 12
c = 20
m = 165
x = 31

values = []

for i in range(1, 200):   # large enough to detect repetition
    x = np.mod((a*x + c), m)
    values.append(x)

print(values)

# 1a find period
first = values[0]
period = values.index(first, 1)
print("Period:", period)

# 1b min
print("Min:", min(values))

# 1c max
print("Max:", max(values))


```

```python
# Question 2

# random uniform
import random

random.seed(25)

nums = []
for i in range(20):
    val = random.uniform(1000, 2000)
    nums.append(val)
    print('%07.2f' % val, end=' ')
print()

print("8th number:", round(nums[7], 2))
```

```python
# Question 3

# random normal distribution
import numpy as np

np.random.seed(50)

nums = np.random.normal(loc=625, scale=125, size=20)
print(nums[9])
```
